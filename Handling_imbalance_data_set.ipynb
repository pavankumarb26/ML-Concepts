Handling imbalance data set

```python
import numpy as np
import pandas as pd

credit_card = pd.read_csv('/content/credit_data.csv')
credit_card.head()
```

```python
credit_card['Class'].value_counts()
```

This is highly unbalanced Data set

0 --> Legit Transactions
1 --> Fraudlent Transactions

```python
legit = credit_card[credit_card.Class == 0]
fraud = credit_card[credit_card.Class == 1]
```

```python
print(legit.shape)
print(fraud.shape)
```

Build a sample data set containing Similar distribution of legit and fraudulent transaction

Under-sampling

```python
legit_sample = legit.sample(n=403)
```

```python
print(legit_sample.shape)
```

Concatenate the two dataframes

```python
new_dataset = pd.concat([legit_sample, fraud],axis = 0)
```

```python
new_dataset.head()
```
